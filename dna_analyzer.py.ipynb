```python
import numpy as np

def dna_analiz_laboratuvari(dna_metni):
    """
    Gelişmiş NumPy fonksiyonları kullanarak DNA analizi yapar, 
    istatistikleri hesaplar ve görsel bir laboratuvar raporu sunar.
    """
    # Standartlaştırma ve NumPy Array'ine dönüştürme
    dna_array = np.array(list(dna_metni.upper()))
    toplam_baz = len(dna_array)
    
    # ------------------------------------------------------------------
    # 🔍 1. ADIM: NumPy Vektörel Frekans Sayımı (Döngüsüz Işık Hızında)
    # ------------------------------------------------------------------
    bazlar = ['A', 'T', 'G', 'C']
    # Her baz için True/False matrisi oluşturup np.sum ile adetleri buluyoruz
    sayilar = {baz: int(np.sum(dna_array == baz)) for baz in bazlar}
    
    # ------------------------------------------------------------------
    # 📊 2. ADIM: İstatistiksel Analizler (np.mean & Kalite Skorlaması)
    # ------------------------------------------------------------------
    # np.mean ile G ve C oranını tek hamlede hesaplıyoruz
    gc_orani = np.mean((dna_array == 'G') | (dna_array == 'C')) * 100
    
    # [Ek Özellik]: DNA Stabilite Skoru (Guanin ve Sitozin bağları üçlü olduğu için zinciri güçlü tutar)
    # Eğer GC oranı %40-%60 arasındaysa bu genetik olarak "Mükemmel Stabilite" demektir.
    stabilite_skoru = "YÜKSEK 🟢" if 40 <= gc_orani <= 60 else "DÜZENSİZ 🔴"

    # ------------------------------------------------------------------
    # 🎨 3. ADIM: Göz Zevkine Uygun Ekran Çıktıları ve Tasarım (Printler)
    # ------------------------------------------------------------------
    print("\n" + "="*60)
    print("🔬  AI-POWERED BIOINTELLIGENCE & DNA ANALYTICS LAB  🔬")
    print("="*60)
    print(f"🧬  Analiz Edilen Zincir : {dna_metni[:30]}... ({toplam_baz} Baz)")
    print(f"⏳  İşlem Süresi         : Ölçülemeyecek Kadar Hızlı (NumPy Vektörize)")
    print("-"*60)
    
    # 🌟 İstediğin Yıldızlı Adet Grafiği (Dinamik Görselleştirme)
    print("📊  DNA BAZ DAĞILIM GRAFİĞİ")
    print("-"*60)
    for baz, adet in sayilar.items():
        # Her 2 adet baz için 1 yıldız çizdirerek grafiği dengeliyoruz (Ölçekleme)
        yildiz_sayisi = adet
        grafik_cubugu = "◼" * yildiz_sayisi  # Yıldız yerine daha modern bir blok karakteri
        print(f"   Baz {baz} [{adet:02d}] : {grafik_cubugu}")
    print("-"*60)
    
    # 📈 İleri Seviye İstatistik Raporu
    print("📈  BİYOİNFORMATİK İSTATİSTİK RAPORU")
    print("-"*60)
    print(f"   ▪️ Toplam GC İçeriği Oranı : %{gc_orani:.2f}")
    print(f"   ▪️ Moleküler Stabilite     : {stabilite_skoru}")
    print("-"*60)

    # ------------------------------------------------------------------
    # ⚠️ 4. ADIM: Yeniden Tasarlanmış Gelişmiş If-Else Karar Mekanizması
    # ------------------------------------------------------------------
    print("🚨  MUTASYON VE SAĞLIK KONTROLÜ")
    print("-"*60)
    
    c_adedi = sayilar['C']
    kritik_esik = 15
    
    if c_adedi >= kritik_esik:
        print("   ✅ ANALİZ SONUCU: DNA zinciri tamamen sağlıklı görünüyor.")
        print(f"   ℹ️ Sitozin (C) adedi: {c_adedi} (Kritik eşik olan {kritik_esik} değerinin üzerinde).")
    else:
        print("   ⚠️ SİNYAL UYARISI: Kritik Sitozin (C) Eksikliği!")
        print(f"   🚨 Mevcut C Adeti: {c_adedi} | Kritik Eşik: {kritik_esik}")
        print("   📢 Rapor: Genetik mutasyon riski saptandı. Laboratuvar kontrolü önerilir.")
        
    print("="*60 + "\n")
    
    # Verileri dışarıda da kullanabilmek için rapor sözlüğünü döndürüyoruz
    return {"sayilar": sayilar, "gc_orani": gc_orani}

# --- PROJENİN CANLI TESTİ (HASTA ZİNCİRİ) ---
test_zinciri = "AGCTTTTCATTCTGACTGCAACGGGCAATATGTCTCTGTGTGGATTAAAAAAAGAGTGTCTGATAGCAGC"
analiz_raporu = dna_analiz_laboratuvari(test_zinciri)
```
